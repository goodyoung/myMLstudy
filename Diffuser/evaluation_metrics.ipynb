### CLIP Similarity

```python
# !pip install git+https://github.com/openai/CLIP.git
```

```python
import torch
import clip
from PIL import Image
import os

# CLIP 로드
device = "cuda" if torch.cuda.is_available() else "cpu"
model, preprocess = clip.load("ViT-B/32", device=device)

# 이미지 경로
real_path = './real/dog_023_after.jpg'
predict_path = './predict/predict1.png'

# 이미지 로드 & 전처리
image_real = preprocess(Image.open(real_path)).unsqueeze(0).to(device)
image_predict = preprocess(Image.open(predict_path)).unsqueeze(0).to(device)

# 임베딩 계산
with torch.no_grad():
    image_real_features = model.encode_image(image_real)
    image_predict_features = model.encode_image(image_predict)

# 정규화
image_real_features /= image_real_features.norm(dim=-1, keepdim=True)
image_predict_features /= image_predict_features.norm(dim=-1, keepdim=True)

# 코사인 유사도 계산
similarity = (image_real_features @ image_predict_features.T).item()
print(f"CLIP 이미지 유사도 (cosine): {similarity:.4f}")
```
